```python

# Name: Alejandra Ortega
# File Name: automobile.py
# Description: This program asks the user for information
# about a car and displays the information using classes.
#
# Variables:
# vehicle_type - Stores the type of vehicle
# year - Stores the year of the car
# make - Stores the manufacturer of the car
# model - Stores the model of the car
# doors - Stores the number of doors
# roof - Stores the type of roof
# car - Stores the automobile object


class Vehicle:
    def __init__(self, vehicle_type):
        self.vehicle_type = vehicle_type


class Automobile(Vehicle):
    def __init__(self, year, make, model, doors, roof):
        super().__init__("car")
        self.year = year
        self.make = make
        self.model = model
        self.doors = doors
        self.roof = roof


year = input("Enter the year: ")
make = input("Enter the make: ")
model = input("Enter the model: ")
doors = input("Enter the number of doors (2 or 4): ")
roof = input("Enter the roof type (solid or sun roof): ")

car = Automobile(year, make, model, doors, roof)

print("\nVehicle Information")
print("Vehicle type:", car.vehicle_type)
print("Year:", car.year)
print("Make:", car.make)
print("Model:", car.model)
print("Number of doors:", car.doors)
print("Type of roof:", car.roof)
```
